**欧几里得算法**

```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))
print(gcd(-12,-8))
```

**问**：如果⼀个或两个参数是负数，你的函数会做什么？这种⾏为有意义吗？

**答**：当传入负数参数时，算法依然可以完成循环迭代，但最终返回值为负数。而数学上 GCD 定义为正整数，所以这个返回结果在数学上是无意义的。

**近似欧拉数e**

```python
import math

def euler_limit(n):
    return (1 + 1/n)**n

def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
    return total

def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) >= epsilon:
        if k ==0:
            fact = 1
        else:
            fact *=k
        total += 1.0/fact
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

# 测试
print(euler_infinite_sum(4))
print_euler_sum_table(6)
print("真实e值：", math.exp(1))
```

观察打印表格可以看到：随着项数 k 增大，求和结果不断逼近真实e，但项数有限时只能得到近似值。

**测试整数属性**

```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

def is_power(b,n):
    while n >1:
        if n % b != 0:
            return False
        n = n // b
    return True

# 测试样例
print(is_power_of_2(8))
print(is_power_of_2(7))
print(is_power(3,81))
print(is_power(5,10))
```
