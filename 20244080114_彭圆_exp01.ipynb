```python
#1. 分别计算8 / 3、8 // 3、8 % 3和2 ** 7，输出结果及其类型，说明各运算符的含义。
a = 8/3               # /  浮点除法
b = 8//3              # // 整除（向下取整）
c = 8%3               # %  取余
d = 2**7              # ** 幂运算
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
```

```python
#3. 执行int("123")、float("3.14")和int(3.9)，解释转换结果；捕获int("goat")引发的异常并说明原因。
print(int("123"))         #数字字符串转整数，成功转换为 123
print(float("3.14"))      #数字字符串转浮点数，成功转换为 3.14
print(int(3.9))           #浮点数转整数，直接截断小数部分，结果为 3
try:
    int("goat")           #非数字字符串无法转换为整数，抛出`ValueError`
except ValueError as e:
    print("捕获异常：",e)
```

```python
#4. 对字符串" Python Data Analysis "完成去除首尾空白、大小写转换、切片及单词拆分，再使用join()将单词连接起来。
text = " Python Data Analysis "
s1 = text.strip()        #去除首尾空白
s2 = s1.upper()          #全大写
s3 = s1.lower()          #全小写
s4 = s1[0:6]             #切片
s5 = s1.split()          #按空格拆分单词
s6 = "-".join(s5)        #用-连接单词
print(s1)
print(s2)
print(s3)
print(s4)
print(s5)
print(s6)
```

实验2：函数、条件、循环与递归

```python
#1. 编写score_level(score)函数：90～100分返回“优秀”，80～不足90分返回“良好”，60～不足80分返回“及格”，0～不足60分返回“不及格”。对于超出0～100范围数值，抛出ValueError。
def score_level(score):
    # 检查分数范围
    if not isinstance(score, (int, float)):
        raise ValueError("分数必须为数字")
    if score < 0 or score > 100:
        raise ValueError("分数必须在0~100范围内")
    # 判断等级
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 功能测试
print(f"95分：{score_level(95)}")
print(f"82分：{score_level(82)}")
print(f"70分：{score_level(70)}")
print(f"55分：{score_level(55)}")
# 测试超出范围（取消注释查看异常）
# score_level(105)
```

```python
#2. 分别使用for和while计算1到100的整数和，比较两种实现的结果。
# for循环实现
sum_for = 0
for i in range(1, 101):
    sum_for += i

# while循环实现
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print(f"for循环计算1~100和：{sum_for}")
print(f"while循环计算1~100和：{sum_while}")
print(f"两种方式结果相等：{sum_for == sum_while}")
```
