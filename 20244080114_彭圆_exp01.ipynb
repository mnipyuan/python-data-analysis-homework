实验2：函数、条件、循环与递归

```python
#1. 编写score_level(score)函数：90～100分返回“优秀”，80～不足90分返回“良好”，60～不足80分返回“及格”，0～不足60分返回“不及格”。对于超出0～100范围数值，抛出ValueError。
def score_level(score):
    # 检查分数范围
    if not isinstance(score, (int, float)):
        raise ValueError("分数必须为数字")
    if score < 0 or score > 100:
        raise ValueError("分数必须在0~100范围内")
    # 判断等级
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 功能测试
print(f"95分：{score_level(95)}")
print(f"82分：{score_level(82)}")
print(f"70分：{score_level(70)}")
print(f"55分：{score_level(55)}")

# 测试超出范围
score_level(105)
```

```python
#2. 分别使用for和while计算1到100的整数和，比较两种实现的结果。
# for循环实现
sum_for = 0
for i in range(1, 101):
    sum_for += i

# while循环实现
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print(f"for循环计算1~100和：{sum_for}")
print(f"while循环计算1~100和：{sum_while}")
print(f"两种方式结果相等：{sum_for == sum_while}")
```

```python
#3. 编写递归版和迭代版阶乘函数，约定0! = 1，输入为非负整数。比较n = 0、1、5、10时的结果。
# 迭代版阶乘
def factorial_iter(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# 递归版阶乘
def factorial_recur(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recur(n - 1)

# 测试不同n值
test_n = [0, 1, 5, 10]
for n in test_n:
    iter_res = factorial_iter(n)
    recur_res = factorial_recur(n)
    print(f"n={n} | 迭代版：{iter_res} | 递归版：{recur_res} | 结果一致：{iter_res == recur_res}")
```

```python
#4. 编写summary(values)函数，接收非空数值列表，以元组返回最小值、最大值和平均值，并通过元组解包接收结果。
def summary(values):
    if not values:
        raise ValueError("输入列表不能为空")
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)

# 测试与元组解包
num_list = [85, 58, 92, 76, 60]
min_v, max_v, avg_v = summary(num_list)
print(f"最小值：{min_v}")
print(f"最大值：{max_v}")
print(f"平均值：{avg_v:.2f}")
```

实验4：文件读写、异常处理与模块

```python
#1. 使用os.getcwd()查看工作目录，使用os.path.join()构造相对路径，并使用os.makedirs(..., exist_ok=True)创建数据目录。
import os

# 查看当前工作目录
current_dir = os.getcwd()
print(f"当前工作目录：{current_dir}")

# 构造数据目录路径
data_dir = os.path.join(current_dir, "exp01_data")
print(f"数据目录路径：{data_dir}")

# 创建目录，exist_ok=True表示已存在不报错
os.makedirs(data_dir, exist_ok=True)
print("数据目录创建完成")
```

```python
#2. 使用with open(..., encoding="utf-8")写入上述数据，再读取文件，跳过表头，使用字符串方法解析记录。
file_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

# 文件路径
txt_path = os.path.join(data_dir, "students.txt")

# 写入文件
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(file_content)
print("students.txt写入完成")

# 读取文件，跳过表头
with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

header = lines[0].strip()
data_lines = lines[1:]  # 跳过表头
print(f"表头：{header}")
print(f"数据行数：{len(data_lines)}")
```

```python
#3. 编写parse_student(line)函数：将成绩转换为整数，检查字段数量、姓名是否为空及成绩是否在0～100范围内。无效记录抛出ValueError。
def parse_student(line):
    line = line.strip()
    # 空行判定
    if not line:
        raise ValueError("空行，无有效数据")
    # 按逗号分割
    parts = line.split(",")
    # 检查字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，期望2个，实际{len(parts)}个")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    # 检查姓名
    if not name:
        raise ValueError("姓名为空")
    # 成绩转整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩'{score_str}'无法转换为整数")
    # 检查成绩范围
    if score < 0 or score > 100:
        raise ValueError(f"成绩{score}超出0~100范围")
    return {"name": name, "score": score}

# 简单测试
print("测试有效行：", parse_student("Alice,85"))
```

```python
#4. 按行处理文件，捕获无效记录的异常，记录行号与错误原因，并继续处理后续记录。有效记录保存为字典组成的列表。
valid_records = []
error_records = []

# 逐行处理，行号从2开始（表头是第1行）
for line_no, line in enumerate(data_lines, start=2):
    try:
        record = parse_student(line)
        valid_records.append(record)
    except ValueError as e:
        error_info = f"第{line_no}行：{str(e)}"
        error_records.append(error_info)

print("          有效记录")
for r in valid_records:
    print(r)
print("\n        无效记录错误")
for err in error_records:
    print(err)
```

```python
#5. 将有效记录数、无效记录数、平均分和及格率写入exp01_data/report.txt。分别演示w与a模式，说明二者对已有内容的影响。
# 计算统计值
valid_count = len(valid_records)
invalid_count = len(error_records)
valid_scores = [r["score"] for r in valid_records]
avg_valid = sum(valid_scores) / valid_count
pass_valid_count = sum(1 for s in valid_scores if s >= 60)
pass_valid_rate = pass_valid_count / valid_count

report_path = os.path.join(data_dir, "report.txt")

#w模式：覆盖写入
with open(report_path, "w", encoding="utf-8") as f:
    f.write("    学生数据统计报告    \n")
    f.write(f"有效记录数：{valid_count}\n")
    f.write(f"无效记录数：{invalid_count}\n")
    f.write(f"平均分：{avg_valid:.2f}\n")
    f.write(f"及格率：{pass_valid_rate:.1%}\n")
print("w模式写入完成，原有内容已被覆盖")

#a模式：追加写入
with open(report_path, "a", encoding="utf-8") as f:
    f.write("\n    追加内容：a模式测试    \n")
    f.write("这行是通过a模式追加的内容，不会覆盖之前的内容\n")
print("a模式追加完成")

# 读取查看最终内容
with open(report_path, "r", encoding="utf-8") as f:
    print("\nreport.txt最终内容：")
    print(f.read())
```

实验5：类、继承与对象复制

```python
#1. 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()和__str__()方法。
class Student:
    def __init__(self, name, score):
        # 初始化属性
        self.name = name
        self.score = score
    
    def is_passed(self):
        # 判断是否及格
        return self.score >= 60
    
    def __str__(self):
        # 字符串表示
        return f"Student(姓名：{self.name}，成绩：{self.score})"

# 测试
s = Student("测试同学", 88)
print(s)
print("是否及格：", s.is_passed())
```

```python
#2. 为实验4中的每条有效记录创建一个Student对象，输出对象信息和及格状态。
print("=== 所有有效学生对象 ===")
student_objects = []
for record in valid_records:
    stu = Student(record["name"], record["score"])
    student_objects.append(stu)
    print(f"{stu} | 及格状态：{'及格' if stu.is_passed() else '不及格'}")
```

```python
#3. 定义继承自Student的GraduateStudent类，增加research_topic属性，通过super().__init__()初始化父类属性，并重写__str__()。
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 调用父类初始化
        super().__init__(name, score)
        # 新增属性
        self.research_topic = research_topic
    
    def __str__(self):
        # 重写__str__
        return f"GraduateStudent(姓名：{self.name}，成绩：{self.score}，研究方向：{self.research_topic})"

# 测试
grad = GraduateStudent("张三", 91, "大数据分析")
print(grad)
print("是否及格：", grad.is_passed())
```

```python
#4. 构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分别进行直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，比较结果。每次比较都从新建的原始对象开始。
import copy

print(" 1. 直接赋值 ")
orig1 = {"name": "Alice", "scores": [85, 90]}
o1 = orig1
o1["scores"][0] = 99
print(f"原始对象：{orig1}")
print(f"赋值对象：{o1}")
print("结论：直接赋值是引用同一对象，修改内部列表，原始对象同步改变")

print("\n 2. copy.copy() 浅拷贝 ")
orig2 = {"name": "Alice", "scores": [85, 90]}
o2 = copy.copy(orig2)
o2["scores"][0] = 100
print(f"原始对象：{orig2}")
print(f"浅拷贝对象：{o2}")
print("结论：浅拷贝只复制外层字典，内部嵌套列表仍是同一引用，修改内部列表原始对象会变")

print("\n 3. copy.deepcopy() 深拷贝 ")
orig3 = {"name": "Alice", "scores": [85, 90]}
o3 = copy.deepcopy(orig3)
o3["scores"][0] = 60
print(f"原始对象：{orig3}")
print(f"深拷贝对象：{o3}")
print("结论：深拷贝递归复制所有嵌套对象，修改副本完全不影响原始对象")
```
