# 实验1：正则表达式的基础匹配

### 任务 1： 
在"cat on mat; later: cats and dogs"上使⽤正则表达式，分别匹配⽂本中的数字 \d+、\s+ 和 \bcat\b。

```python
import re

text1 = "cat on mat; later: cats and dogs"

# 分别执行匹配
res_digit = re.findall(r"\d+", text1)
res_space = re.findall(r"\s+", text1)
res_cat = re.findall(r"\bcat\b", text1)

print("\\d+ 匹配数字结果:", res_digit)
print("\\s+ 匹配空白结果:", res_space)
print("\\bcat\\b 匹配完整cat结果:", res_cat)
```

匹配规则：
- `\d+`：匹配 1 个或多个数字
- `\s+`：匹配 1 个或多个空白字符（空格、制表符等）
- `\bcat\b`：匹配完整单词`cat`，`\b`是单词边界，不会匹配`cats`里的 cat

### 任务2：
在"cat dog bird CAT"上使⽤cat|dog。⽤re.findall(r"a|aa|aaa","aaaa")和re.findall(r"aaa|aa|a", "aaaa")展⽰差异。

```python
text2 = "cat dog bird CAT"
res1 = re.findall(r"cat|dog", text2)
print("cat|dog 匹配结果:", res1)

# 展示匹配顺序差异
text3 = "aaaa"
res2 = re.findall(r"a|aa|aaa", text3)
res3 = re.findall(r"aaa|aa|a", text3)

print("a|aa|aaa 匹配结果:", res2)
print("aaa|aa|a 匹配结果:", res3)
```

`cat|dog`匹配 cat 或 dog；正则多选分支是**左优先匹配**，前面的模式匹配成功就不会尝试后面的模式，因此顺序不同结果不同。

###   任务3：
 
在 "ct cat caat caaat" 上分别使⽤ ca*t、ca+t、ca?t、ca{2}t 和ca{1,2}t，输出匹配列表并说明各量词的含义。

```python
text4 = "ct cat caat caaat"

res_star = re.findall(r"ca*t", text4)
res_plus = re.findall(r"ca+t", text4)
res_question = re.findall(r"ca?t", text4)
res_n = re.findall(r"ca{2}t", text4)
res_range = re.findall(r"ca{1,2}t", text4)

print("ca*t :", res_star)
print("ca+t :", res_plus)
print("ca?t :", res_question)
print("ca{2}t :", res_n)
print("ca{1,2}t :", res_range)
```

5 种量词含义：

- `ca*t`：`*` 匹配 a 出现 0 次或多次
- `ca+t`：`+` 匹配 a 出现 1 次或多次
- `ca?t`：`?` 匹配 a 出现 0 次或 1 次
- `ca{2}t`：`{2}` 匹配 a 恰好出现 2 次
- `ca{1,2}t`：`{1,2}` 匹配 a 出现 1~2 次

# 实验2：分组、回溯引⽤与替换

### 任务1：
在 "联系：alice@example.com；备选：bob@test.org" 中，⽤捕获组提取每个邮箱地址的⽤⼾名和域名。

```python
text5 = "联系:alice@example.com;备选:bob@test.org"
pattern_email = r"(\w+)@([\w.]+)"
matches = re.findall(pattern_email, text5)

print("提取结果(用户名, 域名):")
for user, domain in matches:
    print(f"  用户名：{user}，域名：{domain}")
```

用两个捕获组分别提取 @前的用户名和 @后的域名，`(\w+)@([\w.]+)`中第 1 组是用户名，第 2 组是域名。

### 任务2：
 
在 "cat cat, dog cat, bird bird" 中，⽤后向引⽤找出连续重复的单词。

```python
text6 = "cat cat, dog cat, bird bird"
pattern_dup = r"\b(\w+)\s+\1\b"
dup_words = re.findall(pattern_dup, text6)

print("连续重复的单词:", dup_words)
```

`\b(\w+)\s+\1\b` 中，`\1` 是后向引用，代表和第 1 个捕获组完全相同的内容，以此匹配连续重复的单词。

### 任务3：
将 "2024-05-01\n2024/05/02\n2024.05.03" 中的⽇期从“年⽉⽇”形式转换为“⽉⽇年”形式，保留原来的⾏数。可⽤ re.sub()、捕获组和替换表达式完成；展⽰转换前后⽂本。

```python
text7 = """2024-05-01\n2024/05/02\n2024.05.03"""

pattern_date = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
new_text = re.sub(pattern_date, r"\2-\3-\1", text7)

print("转换前")
print(text7)
print("\n转换后")
print(new_text)
```

用捕获组分别提取年、月、日，替换时调换三组顺序；正则同时支持`-` `/` `.`三种分隔符。

### 任务4：
将 "联系我：123-456-7890 或 987.654.3210" 中的电话号码统⼀为 (123)456-7890 形式。

```python
text8 = "联系我:123-456-7890 或 987.654.3210"
pattern_phone = r"(\d{3})[-.](\d{3})[-.](\d{4})"
new_phone = re.sub(pattern_phone, r"(\1) \2-\3", text8)

print("转换前:", text8)
print("转换后:", new_phone)
```

把`-`或`.`分隔的电话，统一为`(123) 456-7890`格式，用捕获组提取三段数字后重组。

# 实验3：使⽤ BeautifulSoup 解析 HTML

### 任务1：
⽤BeautifulSoup(html_doc, "html.parser")构建解析树，输出⻚⾯标题的⽂本、h1的标签名与⽂本。

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

# 构建解析树
soup = BeautifulSoup(html_doc, "html.parser")

# 提取标题文本
title_text = soup.title.get_text()
print("页面标题文本:", title_text)

# 提取h1标签名和文本
h1_tag = soup.h1
print("h1标签名:", h1_tag.name)
print("h1文本内容:", h1_tag.get_text())
```

### 任务2：
使⽤find_all()找出所有class="course"的段落和所有链接；逐项输出链接⽂字与href属性。

```python
# 查找所有 class=course 的段落
course_paras = soup.find_all("p", class_="course")
print("所有course段落：")
for p in course_paras:
    print(" -", p.get_text())

# 查找所有链接
links = soup.find_all("a")
print("\n所有链接信息：")
for link in links:
    print(f"  文字：{link.get_text()}，链接：{link['href']}")
```

### 任务3：
⽤ get_text() 提取⻚⾯可⻅⽂本；⽐较整个⽂档、单个段落和单个链接的⽂本结果

```python
# 整个文档的文本
all_text = soup.get_text()
# 单个段落的文本
single_p_text = soup.p.get_text()
# 单个链接的文本
single_a_text = soup.a.get_text()

print("【整个文档文本】")
print(repr(all_text))  # repr可以显示换行、空格等不可见字符
print("\n【单个段落文本】")
print(single_p_text)
print("\n【单个链接文本】")
print(single_a_text)
```

`get_text()`会去除 HTML 标签，只提取纯文本；对比整个文档、单个段落、单个链接的提取结果差异。
